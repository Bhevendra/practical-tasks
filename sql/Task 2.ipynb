# SQL Data Engineering Task (1 Hour Challenge)

**Difficulty: 7/10**

**Domain:** Ecommerce Analytics

**Focus Areas:** Joins, Window Functions, Aggregations, Data Quality, CTEs, Business Logic, Performance Thinking

You have these Silver Layer tables:

* `clicked_items`
* `customers`
* `order_products`
* `promotions`
* `sales_orders`

---

# Business Scenario

The marketing team launched multiple promotions and wants to understand:

1. Which promotions actually converted into sales.
2. Which customers clicked products but never purchased.
3. Customer buying behavior after clicking.
4. Revenue leakage caused by discounts/promotions.
5. Top-performing products and customer segments.

You need to create analytical SQL outputs for the business team.

# Task 1 — Customer Funnel Analysis

### Objective

Build a customer funnel from:
`CLICK → ORDER`

### Requirements

Create a query that returns:

| metric                                  | value |
| --------------------------------------- | ----- |
| Total customers who clicked products    |       |
| Customers who placed orders             |       |
| Customers who clicked but never ordered |       |
| Conversion Rate %                       |       |

### Rules

* Use distinct customers.
* Conversion rate = ordering customers / clicking customers * 100
* Use CTEs.

---



# Task 2 — Promotion Effectiveness Analysis

### Objective

Find which promotions generated the highest revenue.

### Requirements

Return:

| promotion_id | promotion_name | total_orders | total_revenue | avg_order_value | rank |

### Rules

* Join promotions with orders.
* Revenue = quantity × price.
* Use:

  * `DENSE_RANK()`
  * aggregation
  * window functions
* Rank promotions by revenue descending.

---



# Task 3 — Top 3 Products Per Customer

### Objective

Identify each customer’s top 3 purchased products.

### Requirements

Return:

| customer_id | product_id | total_quantity | product_rank |

### Rules

* Use:

  * `ROW_NUMBER()` or `RANK()`
  * partition by customer
* Only show top 3.

---




# Task 4 — Click to Purchase Time Analysis

### Objective

Measure customer behavior delay.

### Requirements

Find:

* Average time between first click and first purchase.
* Fastest converting customer.
* Slowest converting customer.

### Rules

* Use timestamp/date columns.
* Use `DATEDIFF`.
* Ignore customers who never purchased.

---



# Task 5 — Revenue Leakage Investigation

### Objective

Find revenue loss due to discounts/promotions.

### Requirements

Return:

| month | gross_revenue | discount_amount | net_revenue | leakage_percentage |

### Formula

* Gross Revenue = before discount
* Net Revenue = after discount
* Leakage % = discount / gross * 100

### Rules

* Aggregate monthly.
* Use date formatting functions.

---



# Task 6 — Data Quality Checks

### Objective

Identify bad records in Silver Layer.

### Requirements

Write SQL queries to detect:

1. Duplicate orders
2. Customers with null IDs
3. Orders without products
4. Negative quantities/prices
5. Promotions applied after expiry date
